After doing so, we can see that there are some values in the 2020 dataframe that mess up out interpretation of data. There are abnormally large outliers that might have been an error when they were put in the database, and since they only mess up the visualization of data, we can get rid of them.

```python
data_total_us_2019 = df_mangoes2019[df_mangoes2019['region'] == 'TotalUS']
data_albany_2019 = df_mangoes2019[df_mangoes2019['region'] == 'Albany']

avg_price_total_us_2019 = data_total_us_2019['AveragePrice'].mean()
avg_price_albany_2019 = data_albany_2019['AveragePrice'].mean()


print(f"AveragePrice totalUS 2019: {avg_price_total_us_2019:.2f}")
print(f"AveragePrice Albany 2019:  {avg_price_albany_2019:.2f}")
```

We see that in all scenarios mangoes are more expensive in 2019 than in 2020. We didn't compare prices for special seasons of the years because we wanted to make an conclusion about the whole years and not only seasons. So it was just not necessary.

```python

from scipy import stats


prices_2019 = df_mangoes2019['AveragePrice']
prices_2020 = filtered_df_mangoes2020['AveragePrice']


t_stat, p_value = stats.ttest_ind(prices_2019, prices_2020)


alpha = 0.05 

if p_value < alpha:
    print("The difference in average prices between 2019 and 2020 is statistically significant.")
else:
    print("The difference in average prices between 2019 and 2020 is not statistically significant.")

```

Last but not least we also checked the statistically significance of the average prices in 2019 & 2020. We see that there is siginifact diffrence and our conlcusion is equiped with strong arguments.

### <span style="color:green">2. What type of mangoes (organic or conventional) are more profitable?</span>

For this second question, the first thing we did was to fix the dataframe with the data from all the years from 2017 to 2020 because as we saw in the excersise before there are some errors with the 2020 data. 

So first we droped the rows with the year 2020 from the "df_total_mangoes" dataframe.


```python
df_total_mangoes_filtered = df_total_mangoes[df_total_mangoes['year'] != 2020]
```

Then we aded the new 2020 filtered dataframe that we also creted in the excersise above, without the outliers

```python
filtered_total_mangoes = pd.concat([df_total_mangoes_filtered, filtered_df_mangoes2020], ignore_index=True)
print(filtered_total_mangoes)
```

```python
Q1 = filtered_total_mangoes['AveragePrice'].quantile(0.25)
Q3 = filtered_total_mangoes['AveragePrice'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

final_filtered_total_mangoes = filtered_total_mangoes[(filtered_total_mangoes['AveragePrice'] >= lower_bound) &
                                                      (filtered_total_mangoes['AveragePrice'] <= upper_bound)]
final_filtered_total_mangoes.head(4)
```

Here we filtered the whole dataframe to avoid outliers once again.

----------------------------------------------------------------------------------------------------------------------------
After cleaning the data we decided to group this data by region and type, to later on work on extracting values from each region.

### <span style="color:blue">Step 2. Analyzing the data</span>

### Data grouped by region and type

```python
mangoes_grouped_region_type = final_filtered_total_mangoes.groupby(["region", "type"])

region_type_dfs = {}

for (region, type_mangoes), group in mangoes_grouped_region_type:
    if region not in region_type_dfs:
        region_type_dfs[region] = {}
    region_type_dfs[region][type_mangoes] = group


for region, types in region_type_dfs.items():
    print(f"Region {region}:")
    for product_type, region_type_df in types.items():
        print(f"  Type {type_mangoes}:")
        print(region_type_df)
        print()
```

### <span style="color:blue">Step 4. Understanding the data</span>

First we calculated the average profit in for organic and convential mangoes from our filtered dataframe. 

```python
avg_profit_organic = final_filtered_total_mangoes[final_filtered_total_mangoes['type'] == 'organic']['Profitability per unit'].mean()
avg_profit_conventional = final_filtered_total_mangoes[final_filtered_total_mangoes['type'] == 'conventional']['Profitability per unit'].mean()

print(f"Average profit per unit for organic mangos:        {avg_profit_organic:.2f}")
print(f"Average profit per unit for conventional mangos:   {avg_profit_conventional:.2f}")
```

In the next step we did a statical test for validation of our results.

```python
from scipy import stats

organic_profits = final_filtered_total_mangoes[final_filtered_total_mangoes['type'] == 'organic']['Profitability per unit']
conventional_profits = final_filtered_total_mangoes[final_filtered_total_mangoes['type'] == 'conventional']['Profitability per unit']

t_stat, p_value = stats.ttest_ind(organic_profits, conventional_profits)

if p_value < 0.05:
    print("The difference in profitability between organic and conventional mangos is statistically significant.")
else:
    print("The difference in profitability between organic and conventional mangos is not statistically significant.")
```

Now we wanted to visualise the results to show the diffrence. 

```python
sns.boxplot(x="type", y="Profitability per unit", data=final_filtered_total_mangoes)
plt.xlabel('Mango Type')
plt.ylabel('Profitability per unit')
plt.title('Comparison of Profitability of Organic and Conventional Mangos')
plt.show()
```
